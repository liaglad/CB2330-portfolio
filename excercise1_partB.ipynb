# Dot product formula:

$$
\mathbf{a}\cdot\mathbf{b}=\sum_{i=1}^{N}a_i b_i
$$


This formula multiplies the corresponding elements of two vectors and then sums up the resulting products.

```python
# Function for calculating the dot product of two vectors:
def dot_product(a, b):
  if len(a) == len(b):
    sum = 0
    for element_a, element_b in zip(a, b):
      product = element_a * element_b
      sum = sum + product
    return sum
  else:
    return "Vectors are not same length, cannot calculate dot product"

# Trial with two vectors with three elements
a = [2, 5, 4]
b = [7, 3, 8]

dot_product_of_a_b = dot_product(a, b)
print("The dot product of the vectors are:", dot_product_of_a_b)
```

```python
# Trials with vectors of different lengths then a and b
c = [4, 6, 2, 8]
d = [3, 6, 4, 9]

dot_product_of_c_d = dot_product(c, d)
print("The dot product of the vectors are:", dot_product_of_c_d)
```

```python
# Trial with vectors of different length
e = [4,2]
f = [8,5,4]

dot_product_of_e_f = dot_product(e, f)
print("The dot product of the vectors are:", dot_product_of_e_f)
```
